```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries loaded successfully!")
```

```python
df_raw = pd.read_csv("coffee_health_v3_cwk1.csv")
df_clean = df_raw.copy()

print("Dataset loaded successfully!")
print("Rows and columns:", df_raw.shape)
```

```python
df_raw = df_clean.copy()

print("Dataset loaded successfully!")
print("Dataset size:", df_raw.shape)

display(df_raw.head())
```

```python
print("DATASET INFORMATION")
print("===================")

df_raw.info()

print("\nCOLUMN NAMES")
print(df_raw.columns.tolist())

print("\nDESCRIPTIVE STATISTICS")
display(df_raw.describe(include="all").T)
```

```python
print("SELF-RATED HEALTH DISTRIBUTION")

print(df_raw["SelfRatedHealth"].value_counts ())

print("PERCENTAGE DISTRIBUTION")

print(df_raw["SelfRatedHealth"].value_counts(normalize=True) * 100)
```

```python
print("MISSING VAlUES ANALYSIS")

missing_count = df_raw.isnull().sum()

missing_percent = (df_raw.isnull().mean() * 100).round(2)

missing_table = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percent


})

display(missing_table)
```

```python
print("DUPLICATE RECORDS ANALYSIS")

print("Total duplicate rows:")
print(df_raw.duplicated().sum())

print("Duplicate IDs:")
print(df_raw["ID"].duplicated().sum())
```

```python
print("REMOVING DUPLICATE RECORDS")

df_clean = df_raw.drop_duplicates().copy()

print ("Original rows:", len(df_raw))
print("Rows after cleaning:", len(df_clean))
print("Duplicates remaining:", df_clean.duplicated().sum())
```

```python
print("MISSING VALUES AFTER REMOVING DUPLICATES")

print(df_clean.isnull().sum().sort_values(ascending=False))
```

```python
print("HANDLING MISSING VALUES")


numeric_cols = df_clean.select_dtypes(include="number").columns.drop("ID")

for col in numeric_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())
categorical_cols = df_clean.select_dtypes(include="object").columns

for col in categorical_cols:
  df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

print("Missing valuea remaining:")
print(df_clean.isnull(). sum())
```

```python
print("CHECKING FOR OUTLIERS")

columns_to_check = [
    "Age",
    "BMI",
    "Daily Coffees",
    "Caffeine Intake",
    "Avg Resting Heart Rate",
    "Avg Sleep Hours Per Night"
]

for col in columns_to_check:
  print("\n", col)
  print("Minimum:",df_clean[col].min())
  print("Maximum:", df_clean[col].max())
  print("Median:", df_clean[col].median())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
fig,axes = plt.subplots(1,2, figsize=(12, 5))

sns.boxplot(y=df_clean["Age"], ax=axes[0])
axes[0].set_title("Age before Outlier Cleaning")

sns.boxplot(y=df_clean["BMI"], ax=axes[1])
axes[1].set_title("BMI Before Outlier Cleaning")

plt.tight_layout()
plt.show()
```

```python
print("AGE ANOMALIES")
print("Ages above 110:", (df_clean["Age"]> 110).sum())

print("\nBMI ANOMALIES")
print("BMI values abover 60:", (df_clean["BMI"] > 60).sum())

print("\n EXAMPLE RECORDS")
display(df_clean.loc[
    (df_clean["Age"] > 110) |
    (df_clean["BMI"] > 60),
    ["ID", "Age", "BMI"]

  ].head(15))
```

```python

print("CLEANING AGE ANOMALIES")

age_median = df_clean.loc[
    df_clean["Age"].between(18, 110), "Age"
].median()

df_clean.loc[
    ~df_clean["Age"].between(18, 110), "Age"
] = age_median

print("Median age used:", age_median)
print("Ages above 110 remaining:", (df_clean["Age"] > 110).sum())
```

```python

print("CHECKING BMI ANOMALIES")

print("BMI values above 60:",
      (df_clean["BMI"] > 60).sum())

display(
    df_clean.loc[
        df_clean["BMI"] > 60,
        ["ID", "BMI"]
    ].head(15)
)
```

```python

print("CLEANING BMI ANOMALIES")

bmi_mask = df_clean["BMI"] > 60

df_clean.loc[bmi_mask, "BMI"] = (
    df_clean.loc[bmi_mask, "BMI"] / 10
)

print("BMI values corrected:", bmi_mask.sum())
print("BMI values above 60 remaining:",
      (df_clean["BMI"] > 60).sum())

print("New BMI minimum:", df_clean["BMI"].min())
print("New BMI maximum:", df_clean["BMI"].max())
```

```python

print("SAVING CLEANED DATASET")

df_clean.to_csv("coffee_health_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
print("Final dataset shape:", df_clean.shape)
print("Missing values remaining:",
      df_clean.isnull().sum().sum())
```

```python

print("COFFEE CONSUMPTION AND HEALTH")

health_order = [
    "Poor", "Fair", "Good",
    "Very Good", "Excellent"
]

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df_clean,
    x="SelfRatedHealth",
    y="Daily Coffees",
    order=health_order
)

plt.title("Daily Coffee Consumption by Health Rating")
plt.xlabel("Self-Rated Health")
plt.ylabel("Daily Coffees")
plt.tight_layout()
plt.show()

print("AVERAGE DAILY COFFEES BY HEALTH RATING")

display(
    df_clean.groupby("SelfRatedHealth")["Daily Coffees"]
    .agg(["mean", "median", "count"])
    .reindex(health_order)
    .round(2)
)
```

```python

print("STRESS LEVEL AND SELF-RATED HEALTH")

health_order = [
    "Poor", "Fair", "Good",
    "Very Good", "Excellent"
]

stress_order = ["Low", "Medium", "High"]

# Convert stress categories into numbers
df_clean["StressNumeric"] = df_clean["Stress Level"].map({
    "Low": 1,
    "Medium": 2,
    "High": 3
})

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df_clean,
    x="SelfRatedHealth",
    y="StressNumeric",
    order=health_order
)

plt.title("Stress Level by Self-Rated Health")
plt.xlabel("Self-Rated Health")
plt.ylabel("Stress Level")
plt.yticks([1, 2, 3], stress_order)
plt.tight_layout()
plt.show()

display(
    df_clean.groupby("SelfRatedHealth")["StressNumeric"]
    .agg(["mean", "median"])
    .reindex(health_order)
    .round(2)
)
```

```python

print("SPEARMAN CORRELATION ANALYSIS")

df_corr = df_clean.copy()

df_corr["HealthNumeric"] = df_corr["SelfRatedHealth"].map({
    "Poor": 1,
    "Fair": 2,
    "Good": 3,
    "Very Good": 4,
    "Excellent": 5
})

columns = [
    "Age",
    "BMI",
    "Daily Coffees",
    "Caffeine Intake",
    "Avg Resting Heart Rate",
    "Avg Sleep Hours Per Night",
    "StressNumeric",
    "HealthNumeric"
]

corr = df_corr[columns].corr(method="spearman")

plt.figure(figsize=(11, 8))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f"
)

plt.title("Spearman Correlation Heatmap")
plt.tight_layout()
plt.show()

print("CORRELATION WITH SELF-RATED HEALTH")
display(
    corr["HealthNumeric"]
    .sort_values(ascending=False)
)
```

```python

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
import seaborn as sns

print("K-MEANS CLUSTERING ANALYSIS")

cluster_columns = [
    "Age",
    "BMI",
    "Daily Coffees",
    "Caffeine Intake",
    "Avg Resting Heart Rate",
    "Avg Sleep Hours Per Night",
    "StressNumeric"
]

X = df_clean[cluster_columns].dropna()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Compare different numbers of clusters
scores = {}

for k in range(2, 7):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    labels = model.fit_predict(X_scaled)
    scores[k] = silhouette_score(
        X_scaled,
        labels,
        sample_size=min(2000, len(X)),
        random_state=42
    )

print("Silhouette scores:")
print(scores)

best_k = max(scores, key=scores.get)
print("Best number of clusters:", best_k)

model = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

df_clean.loc[X.index, "Cluster"] = model.fit_predict(X_scaled)

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df_clean.loc[X.index],
    x="BMI",
    y="Avg Sleep Hours Per Night",
    hue="Cluster",
    palette="viridis",
    alpha=0.5
)

plt.title("Health and Lifestyle Clusters")
plt.tight_layout()
plt.show()

print("CLUSTER PROFILES")

display(
    df_clean.groupby("Cluster")[cluster_columns]
    .mean()
    .round(2)
)
```

```python

print("CLUSTER HEALTH COMPARISON")

cluster_health = pd.crosstab(
    df_clean["Cluster"],
    df_clean["SelfRatedHealth"],
    normalize="index"
) * 100

health_order = [
    "Poor", "Fair", "Good",
    "Very Good", "Excellent"
]

cluster_health = cluster_health.reindex(
    columns=health_order,
    fill_value=0
)

display(cluster_health.round(2))

cluster_health.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 6),
    colormap="viridis"
)

plt.title("Self-Rated Health Distribution by Cluster")
plt.xlabel("Cluster")
plt.ylabel("Percentage of Participants")
plt.xticks(rotation=0)
plt.legend(
    title="Self-Rated Health",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)
plt.tight_layout()
plt.show()

print("CLUSTER SIZES")
print(df_clean["Cluster"].value_counts())
```
